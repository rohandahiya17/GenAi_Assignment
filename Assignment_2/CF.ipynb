# Task - 1 (Discount Rules)


```python
Order_amount=input("Enter the order amount: ")
try:
    Order_amount=float(Order_amount)
    if Order_amount >= 2000:
        Discount_persent=15/100
    elif 1500 <= Order_amount < 2000:
        Discount_persent=10/100
    elif 1000 <= Order_amount < 1500:
        Discount_persent=7/100
    else:
        Discount_persent=0
    discount=Order_amount*Discount_persent
    Subtotal=Order_amount-discount
    print("Subtotal: ",Subtotal)
except ValueError:
    print("Invalid input. Please enter a valid number.")


# Task - 1 (Extra) 
tax_rate = 0.05
tax_amount = Subtotal * tax_rate
Total_amount = Subtotal + tax_amount
print("Tax Amount: ", tax_amount)
print("Total Amount after Tax: ", Total_amount)
```

# Task - 2 (Process Multiple Orders) 

```python
Order_amount=[1200, 2500, 800, 1750, 3000]
Total_revenue=0
Discount_count=0
for amount in Order_amount:
    if amount >= 2000:
        Discount_persent=15/100
    elif 1500 <= amount < 2000:
        Discount_persent=10/100
    elif 1000 <= amount < 1500:
        Discount_persent=7/100
    else:
        Discount_persent=0
    discount=amount*Discount_persent
    Final_amount=amount-discount
    print(f"Order_amount:{amount}, Discount: {discount}, Final Amount: {Final_amount}")
    Total_revenue+=Final_amount
    print(f"Total Revenue: {Total_revenue}")


# Task - 2 (Extra)
    if Discount_persent > 0:
        Discount_count += 1
        print(f"Number of orders with discount: {Discount_count}")
```

# Task - 3 (User Menu)

```python
Orders = []

while True:
    print("\nMenu Options:")
    print("1 - Add Order")
    print("2 - View Orders and Total After Discounts")
    print("Q - Quit")
    
    choice = input("Enter your choice (1, 2, Q): ")
    
    if choice == '1':
        amount = float(input("Enter the order amount: "))
        Orders.append(amount)

    elif choice == '2':
        if len(Orders) == 0:
            print("No orders recorded yet.")
        else:
            print("\n--- Order Breakdown ---")
            print(f"All Orders: {Orders}")
            
            for order in Orders:
                if order >= 2000:
                    discount_percent = 0.15
                elif 1500 <= order < 2000:
                    discount_percent = 0.10
                elif 1000 <= order < 1500:
                    discount_percent = 0.07
                else:
                    discount_percent = 0
                    
                discount = order * discount_percent
                final_amount = order - discount
                print(f"  Amount: {order} , Discount: {discount} , Final: {final_amount}")
    elif choice == 'Q' or choice == 'q':
        print("Exiting the program. Goodbye!")
        break
    else:
        print("Invalid choice. Please select a valid option.")
        continue
```

# Task - 4 (Break & Continue)

```python
Daily_sales = [200,150,0,400,50,-1,300]
print("Daily_sales: ", Daily_sales)
total_sales = 0
for sale in Daily_sales:
    if sale < 0:
        print("Corrupted data")
        break
    if sale == 0:
        print("No sales today")
        continue
    else:
        print(f"Processing sale amount: {sale}")
        total_sales += sale
print(f"Total sales for the week: {total_sales}")
```
