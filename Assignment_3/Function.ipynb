# Task - 1 (Basic Function)

```python
def apply_discount(price, discount_percent = 5):
    if discount_percent > 60:
        discount_percent = 60
    discount_amount = price * (discount_percent / 100)
    discounted_price = price - discount_amount
    return discounted_price

# test cases
print(apply_discount(1000, 10))
print(apply_discount(500))
print(apply_discount(10000, 70))  
```

# Task - 2 (Recusrive Function)

```python
def factorial_recursive(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."
    elif n == 0 or n == 1:
        return 1
    else:
        return n * factorial_recursive(n - 1)

# Test cases
print(factorial_recursive(5))
print(factorial_recursive(0))
print(factorial_recursive(-3))
```

# Task - 3 (Lambda Function)

```python
GST=lambda price: price+(price * 0.18)
print(GST(1000))

# Task - 3 (Extra)
After_Discount=lambda price,discount: price - (price * discount / 100)*0.18
print(After_Discount(1000, 10))
```

# Task - 4 (Map Function)

```python
price=[100,250,400,1200,50,2000,850]
GST=lambda price: price+(price * 0.18)
Price_after_gst=list(map(GST,price))
print(f"Original price: {price}")
print(f"Price after GST: {Price_after_gst}")
```

# Task - 5 (Filter Function)

```python
price=[100,250,400,1200,50,2000,850]
Greater_than_500=filter(lambda price: price>500,price)
Less_than_or_equal_to_500=filter(lambda price: price <= 500,price)
print(f"Prices greater than 500: {list(Greater_than_500)}")
print(f"Prices less than or equal to 500: {list(Less_than_or_equal_to_500)}")
```

# Task - 6 (Combined Function)

```python
def process_prices(prices):
    Discount_price=list(map(lambda price: price - (price * 0.10),prices))
    Filtered_prices=list(filter(lambda Discounted_price: Discounted_price > 300,Discount_price))
    return Discount_price, Filtered_prices

# Test cases
process_prices([100, 500, 900, 50, 750])
```

# Task - 7 (Mini Project)

```python
def add_price(price_list, price):
  price_list.append(price)
  
def get_average_price(prices_list):
  if len(prices_list) == 0:
    return 0
  return sum(prices_list) / len(prices_list)


def get_max_price(prices_list):
  if len(prices_list) == 0:
    return None
  return max(prices_list)


def main():
  prices_list = []
  while True:
    print("Menu:")
    print("1 - Add price")
    print("2 - Show average price")
    print("3 - Show maximum price")
    print("q - Quit")

    choice = input("Enter your choice (1,2,3,q): ")

    if choice == "1":
      try:
        price = float(input("Enter the price to add: "))
        add_price(prices_list, price)
        print(f"Price {price} added.")
      except ValueError:
        print("Invalid input! Please enter a valid number for the price.")
    elif choice == "2":
      average_price = get_average_price(prices_list)
      print(f"Average price: {average_price:.2f}")
    elif choice == "3":
      max_price = get_max_price(prices_list)
      if max_price is None:
        print("Maximum price: No prices added yet.")
      else:
        print(f"Maximum price: {max_price}")
    elif choice == "q":
      print("Exiting the program.")
      break
    else:
      print("Invalid choice. Please try again.")


if __name__ == "__main__":
  main()
```
