# Task - 2 (Categories(Sets))

```python
Products_categories=["Fruits","Stationery","Sports","Furniture","Electronics","Transport","Food","Sports"]
Categories_set=set(Products_categories)
Categories_set.add("Dog breed")
print(Categories_set)

category_check=input("Enter the name:")
if category_check in Categories_set:
    print("True")
else:
    print("False")

# Task - 2 (Extra)
total_unique_set=len(Categories_set)
print("Total number of unique set",total_unique_set)
```

# Task - 3 Product Pricing(Dictionaries)

```python
price_dict={"Apple":300,"Book":100,"Bet":5000,"Table":2000,"Laptop":40000,"Car":4000000}
price_dict["Cake"]=550
price_dict["Bet"]=3000
del price_dict["Car"]
print(price_dict)

Total_price=sum(price_dict.values())
Total_number=len(price_dict)
if Total_number > 0:
    avg_price=Total_price/Total_number
    print("Average proce of all products",avg_price)

# Task - 3 (Extra)
Max_product=max(price_dict, key=price_dict.get) 
Max_price=price_dict[Max_product]
Min_product=min(price_dict, key=price_dict.get) 
Min_price=price_dict[Min_product]
print(f"Most Expensive Product: {Max_product} (₹{Max_price})")
print(f"Least Expensive Product: {Min_product} (₹{Min_price})")
```

# Task - 4 (Combined operations)

```python
zipped=list(zip(Products,Products_categories))
print(zipped)
Catalog=[(name,price_dict[name],category)
        for name,category in zipped
        if name in price_dict
        ]
print(Catalog)

category_to_products={}
for name , price , category in Catalog:
    category_to_products.setdefault(category,[]).append(name)
print(category_to_products)

if category_to_products:
    max_category = max(category_to_products, key=lambda cat: len(category_to_products[cat]))
    print(f"Category with most products ({max_category}): {category_to_products[max_category]}")
```
