# Task - 1 (Write file)

```python
sales=[1200,450,980,1500,3000]
with open("sales_data.txt", "w") as sale:
    for price in sales:
        sale.write(f"{price}\n")
with open("sales_data.txt", "r") as sale:
    content = sale.read()
    print(content)
with open("sales_data_comma.txt","w") as sale:
    sale.write(",".join(map(str, sales)))
```

# Task - 2 (Read File)

```python
print("Reading the sales_data.txt file and printing it .")
with open("sales_data.txt","r")as sale:
    content = sale.read()
print(content)
print("Read the first line & all lines using .readline() methods")
with open("sales_data.txt","r")as sale:
    first_line = sale.readline()
    print(first_line)
with open("sales_data.txt", "r") as sale:
    sales_integers = [int(line.strip()) for line in sale.readlines()]
    print(sales_integers)
```

# Task - 3 (Append new sales)

```python
new_sales=[1500,2500,1700]
with open("sales_data.txt", "a") as sale:
    for price in new_sales:
        sale.write(f"{price}\n")

# Task- 3 (Extra)
with open("sales_data.txt", "r") as sale:
    total_lines = sale.readlines()
    print(f"Total number of lines in the file: {len(total_lines)}")
```

# Task - 4 (Summary Report)

```python
with open("sales_data.txt","r") as sale:
    sales_integers = [int(line.strip()) for line in sale.readlines()]
total_sales = sum(sales_integers)
highest_sale = max(sales_integers)
lowest_sale = min(sales_integers)
average_sale = total_sales / len(sales_integers)
print(f"Total Sales: {total_sales}, Highest Sale: {highest_sale}, Lowest Sale: {lowest_sale}, Average Sale: {average_sale}")
```

# Task - 5 (User Input)

```python
List_pr=[]
for product in range(3):
    name = input("Enter the product name: ")
    price = float(input("Enter the product price: "))
    List_pr.append((name, price))
with open("product.txt", "w") as product_file:
    for name, price in List_pr:
        product_file.write(f"{name} | {price}\n")
with open("product.txt", "r") as product_file:
    content = product_file.read()
    print(content)
```

# Task - 7 (Mini Project)

```python
prices={"Mouse":500,"Keyboard":800,"Monitor":7000,"Pendrive":400,"Camera":5000}
discounted_precentage=float(input("Enter discount percentage:"))
discount_list=[]
with open("discount_report.txt","w")as report:
    report.write("Product|Orignal Price|Discounted Price\n")
    for product,orignal_price in prices.items():
        discount_amount=orignal_price*(discounted_precentage/100)
        discount=orignal_price-discount_amount
        discount_list.append(discount)
        report.write(f"{product}|{orignal_price}|{discount}\n")
with open ("discount_report.txt","r") as dr:
    discount_read=dr.read()
    print(discount_read)

# Task - 7 (Extra)
total_item=len(prices)
Average_discounted_price=sum(discount_list)/total_item
print(f"Total Item:{total_item},\nAverage Discounted price:{Average_discounted_price}")
```
